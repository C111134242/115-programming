# Range

Python:
- 懶惰是一種美德
- 由 0 開頭
- 包含頭，不包含尾

```python
#區間，由 0 開始，到 5-1 結束 (0是預設值，可省略)
print(list(range(5)))
#區間，由 0 開始，到 5-1 結束
print(list(range(0,5)))
#區間，由 1 開始，到 5-1 結束
print(list(range(1,5)))
#區間，由 1 開始，到 5-1 結束，每次跳 2 個
print(list(range(1,5,2)))

print(list(range(2,101,2)))
```

# 九九乘法表

1\*1=1, 1\*2=2, 1\*3=3, ......  
2\*1=2, 2\*2=4, 2\*3=6, ......  
3\*1=3, 3\*2=6, 3\*3=9, ......  



```python
#思考過程記錄
j=1
print(f'1*{j}={1*j}',end=',')
j=2
print(f'1*{j}={1*j}',end=',')
j=3
print(f'1*{j}={1*j}',end=',')
```

```python
i=1
for j in range(1,6):
    print(f'{i}*{j}={i*j}',end=',')
print()

i=2
for j in range(1,6):
    print(f'{i}*{j}={i*j}',end=',')
print()

i=3
for j in range(1,6):
    print(f'{i}*{j}={i*j}',end=',')
print()
```

```python
for i in range(1,10):
    for j in range(1,10):
        print(f'{i}*{j}={i*j}',end=',')
    print()
```

# 練習：印出質數

質數：一個數字 $n$ ,除了自己本身，以及1以外，沒有任何數可以除盡他，$n$ 就是質數。

```python
n=273
i=2
if n % 2==0:
    print(f'{273}不是質數')
i=3
if n % 3==0:
    print(f'{273}不是質數')
i=5
if n % 5==0:
    print(f'{273}不是質數')
```

```python
for i in range(2,273):
    if 273%i==0:
        print(f'{273}不是質數,因為它可以被{i}除盡')
        break #只要數字能被除盡，就跳離迴圈
```

```python
n=273
isPrime=True #假設n是質數
for i in range(2,n):
    if n%i==0:
        print(f'{n}不是質數,因為它可以被{i}除盡')
        isPrime=False #確定了n不是質數
        break #只要數字能被除盡，就跳離迴圈

if isPrime==True: #判斷我們的假設是否還成立
    print(f'{n}是質數')
```
