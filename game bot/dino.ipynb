# 介紹
使用 pyautogui 來完成小恐龍的遊戲機器人。

小恐龍是在 chrome 中使用 `chrome://dino` 啟動。

邏輯:
- 不斷監視畫面
- 讀取某個位置的顏色
- 若此位置出現(83,83,83)的顏色
- 跳起來

```python
import pyautogui as pyg
pyg.sleep(2) # sleep 2 sec. 
pyg,press('space') #啟動

#無窮迴圈
while True:
    im=pyg.screenshot()
    obs=False #假設沒有仙人掌
    for i in range(15):
        c=im.getpixel((300+i,325))
        if c==(83,83,83):
            obs=True
            break
    if obs==True:
        pyg.press('space')
```

```python
# 呼叫 pyg 進行螢幕截圖，並儲存到 im 中
im=pyg.screenshot()
c=im.getpixel((272,322))
print(c)
```
