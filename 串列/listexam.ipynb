# 簡介

串列(list)是可以儲存多個元素(整數、浮點數、或字串)的容器。  

串列中每個元素都用逗號隔開，左右有中刮號[]

```python
# scores 是一個串列，裡面每個元素都是整數。
scores=[78,29,56,80,99]
print(scores)
```

```python
## 索引值

我們可以使用索引值(index)來選取串列中的文字。
注意：索引值由0開始。
```

```python
print(scores[0]) #印出第1個元素
print(scores[1]) #印出第2個元素
```

## 串列內容的個數

len(a) 取得a這個串列中元素的個數。
len是 length(長度)的意思。

```python
print(len(scores))
print(scores[4])  # 取得最後一個元素
print(scores[5-1])  # 取得最後一個元素
print(scores[len(scores)-1])
print(scores[-1])
```

# 新增元素

使用append()新增元素到串列的尾端

```python
print(scores)
scores.append(78)
scores.append(93)
print(scores)
```

```python
print(scores)
scores[2]=70 #將70指定給索引值為2的元素
print(scores)
```

## 走訪串列的三種寫法

```python
print(scores)
for i in range(len(scores)):
    print(scores[i])
```
